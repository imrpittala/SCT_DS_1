## 1. Environment Setup & Library Imports
Importing necessary scientific and visualization packages and establishing uniform styling.


```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Configure styling
sns.set_theme(style='whitegrid', palette='deep')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11

print('Environment initialized successfully.')
```

## 2. Data Acquisition & Structural Profiling
Loading demographic dataset and verifying schema, memory footprint, and absence of missing values.


```
data_path = '../data/dataset.csv'
df = pd.read_csv(data_path)

print(f'Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns')
print('\nData Types and Null Summary:')
print(df.info())
df.head(5)
```

## 3. Parametric & Non-Parametric Summary Statistics
Computing central tendencies, dispersion, and distribution shape metrics for continuous features.


```
continuous_cols = ['Median_Age', 'Life_Expectancy', 'Urban_Population_Pct', 'Total_Population']
stats_summary = df[continuous_cols].describe().T
stats_summary['skewness'] = df[continuous_cols].skew()
stats_summary['kurtosis'] = df[continuous_cols].kurtosis()
stats_summary
```

## 4. Continuous Variable Analysis: Age Distribution & KDE
Visualizing the distribution of global median ages with an annotated histogram and kernel density estimation.


```
plt.figure(figsize=(11, 5))
mean_age = df['Median_Age'].mean()
median_age = df['Median_Age'].median()

ax = sns.histplot(df['Median_Age'], kde=True, bins=16, color='#1f77b4', edgecolor='white', alpha=0.7)
plt.axvline(mean_age, color='#d32f2f', linestyle='--', linewidth=2, label=f'Mean: {mean_age:.1f} yrs')
plt.axvline(median_age, color='#388e3c', linestyle='-.', linewidth=2, label=f'Median: {median_age:.1f} yrs')

plt.title('Global Median Age Distribution with KDE', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Median Age (Years)')
plt.ylabel('Country Count')
plt.legend(frameon=True)
plt.tight_layout()
plt.show()
```

## 5. Categorical Variable Analysis: Regional & Economic Breakdown
Examining national distribution across geographic regions and World Bank income groups using styled horizontal and vertical bar charts.


```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5.5))

# 1. Geographic regions
region_counts = df['Region'].value_counts()
sns.barplot(x=region_counts.values, y=region_counts.index, palette='mako', ax=ax1, edgecolor='black')
ax1.set_title('Countries by Geographic Region', fontweight='bold')
ax1.set_xlabel('Number of Nations')

# 2. Income classification
income_order = ['Low income', 'Lower middle income', 'Upper middle income', 'High income']
income_counts = df['Income_Group'].value_counts().reindex(income_order)
sns.barplot(x=income_counts.index, y=income_counts.values, palette='viridis', ax=ax2, edgecolor='black')
ax2.set_title('Countries by Income Group', fontweight='bold')
ax2.set_xlabel('World Bank Income Tier')
ax2.set_ylabel('Number of Nations')
ax2.tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

## 6. Bivariate Demographic Interactions: Age vs Life Expectancy
Analyzing the relationship between median age and life expectancy across regional economic clusters.


```
plt.figure(figsize=(11, 6))
sns.scatterplot(
    data=df,
    x='Median_Age',
    y='Life_Expectancy',
    hue='Region',
    size='Total_Population',
    sizes=(50, 450),
    alpha=0.85,
    palette='tab10'
)
sns.regplot(
    data=df,
    x='Median_Age',
    y='Life_Expectancy',
    scatter=False,
    color='#37474f',
    line_kws={'linestyle': '--', 'linewidth': 1.8}
)
plt.title('Demographic Transition: Life Expectancy vs Median Age', fontsize=14, fontweight='bold')
plt.xlabel('Median Age (Years)')
plt.ylabel('Life Expectancy at Birth (Years)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)
plt.tight_layout()
plt.show()
```

## 7. Demographic Correlation Heatmap
Computing Pearson correlation coefficients across numerical demographic variables.


```
num_cols = ['Median_Age', 'Age_0_14_Pct', 'Age_15_64_Pct', 'Age_65_Plus_Pct', 'Urban_Population_Pct', 'Life_Expectancy']
corr_matrix = df[num_cols].corr()

plt.figure(figsize=(8, 6.5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.8)
plt.title('Demographic Indicators Correlation Matrix', fontsize=14, fontweight='bold', pad=12)
plt.tight_layout()
plt.show()
```

## 8. Summary of Findings & Takeaways

### Q&A
- **Q: What is the shape of the global median age distribution?**  
  The median age distribution displays a slight positive skew with a central median of approximately 32 years and an interquartile range of 14.5 years, ranging from young demographics in Sub-Saharan Africa (~18-20 years) to super-aged populations in Europe and East Asia (~45-48 years).
- **Q: How does economic tier correlate with demographic longevity?**  
  High-income nations consistently demonstrate median ages exceeding 40 years and life expectancies above 80 years, whereas lower-middle and low-income nations exhibit younger demographic pyramids with higher youth dependency ratios.

### Data Analysis Key Findings
- **Longevity and Age Alignment**: A robust positive correlation (r > 0.85) exists between median population age and life expectancy.
- **Urbanization Transition**: Higher urban population percentages (>75%) correlate strongly with lower youth dependency (Age 0-14 < 20%) and higher elderly shares (Age 65+ > 15%).
- **Regional Disparity**: Sub-Saharan Africa exhibits high population growth rates (>2.3%) and young median ages, while European nations face negative growth rates and demographic aging.
